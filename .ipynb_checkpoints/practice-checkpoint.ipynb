```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([[2.5,2.4],[0.5,0.7],[2.2,2.9],[1.9,2.2],[3.1,3.0],[2.3,2.7],[2,1.6],[1,1.1],[1.5,1.6],[1.1,0.9]])
```

```python
x.shape
```

## ploting

```python
plt.figure(figsize=(8,8))
plt.scatter(x[:,0],x[:,1],marker='o')
plt.axis('equal')
plt.grid(True)
plt.show()
```

## standardize the data

```python
x_standerdize=x-np.mean(x)
```

```python
## plot the standardize data
```

```python
plt.scatter(x_standerdize[:,0],x_standerdize[:,1])
plt.grid(True)
plt.show()
```

## find covarience

```python
covarience_matrix=np.cov(x_standerdize,rowvar=False)
```

```python
covarience_matrix
```

## step4: calculate the eigevalues and eigenvector

```python
eigenvalue,eigenvector=np.linalg.eigh(covarience_matrix)
```

```python
eigenvalue
```

```python
eigenvector
```

## step5: sort eigenvector by eigrnvalues

```python
sorted_value=np.argsort(eigenvalue)[::-1]
```

```python
eigenvalue=eigenvalue[sorted_value]
eigenvalue
```

```python
eigenvector=eigenvector[:,sorted_value]
eigenvector
```

## step6 chosse higest eigenvector

```python
eigenvector_subset=eigenvector[:,0:1]
eigenvector_subset
```

## step7: Transform the data(project into principal componet)

```python
x_reduce=np.dot(x_standerdize,eigenvector_subset)
x_reduce
```
